# 2b · Train delphinID for several detection types (from the app)

Trains a project made in the app with **Whistles and/or clicks** (a delphinID model per detection type and a final forest per block), using Colab's free GPU.

1. In the app: **Run > Train in Google Colab > Make Colab bundle** (the detection frames are made first).
2. Here: **Runtime > Change runtime type > T4 GPU** (optional, usually faster), upload the bundle with the Files panel on the left, and set `BUNDLE` below to its name.
3. **Runtime > Run all.** Each detection type is cross-validated (every event predicted by a model trained without it), the final models and PAMGuard zips are trained, then the final forest per block. The results zip downloads at the end.
4. In the app: **Import results from Colab**.

Settings come from the project (Settings screen in the app). Leaving one event out at a time trains one model per event and type; set **Folds** in the app (e.g. 10) for far fewer models.
